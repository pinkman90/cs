```python
import pytesseract
# Explicitly set the path to the Linux Tesseract binary installed via apt-get
pytesseract.pytesseract.tesseract_cmd = '/usr/bin/tesseract'
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
img = np.ones((100,300,3),dtype=np.uint8)*255
cv2.putText(img,"ABC123", (30, 50), cv2.FONT_HERSHEY_COMPLEX,1.5, (0,0,0),2)
cv2.imwrite("captcha.png", img)
plt.imshow(img)
```

```python
result = pytesseract.image_to_string(img, config="--psm 7")
```

```python
print("Original : ABC123")
print("Result :", result)
```

```python
if result.strip()== "ABC123":
    print("Attack successfull")
else:
    print("Attack Failed")
```
